```
# ✅ Create Streamlit app file

import streamlit as st
from crewai import Agent, Task, Crew, Process, LLM
from crewai_tools import SerperDevTool
import os

# Set environment variables (ensure they are set properly before running)
GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")
SERPER_API_KEY = os.getenv("SERPER_API_KEY")

# AI model
llm = LLM(model="gemini/gemini-1.5-flash", temperature=0.5, verbose=True, api_key=GOOGLE_API_KEY)

# Product catalog
product_catalog = {
    "cleanser": ["Cetaphil Gentle Cleanser - $10", "Simple Micellar Gel Wash - $8"],
    "toner": ["Plum Green Tea Toner - $9", "Minimalist PHA Toner - $11"],
    "moisturizer": ["CeraVe Moisturizing Cream - $15", "Neutrogena Hydro Boost - $13"],
    "serum": ["Minimalist Niacinamide 10% - $12", "The Ordinary Hyaluronic Acid - $14"],
    "sunscreen": ["La Shield SPF 40 - $10", "Re'equil Oxybenzone-Free Sunscreen - $9"]
}

# Streamlit UI
st.title("🧴 AI-Powered Skincare Product Recommender")

skin_type = st.selectbox("Choose your skin type", ["Oily", "Dry", "Combination", "Sensitive", "Normal"])
concerns = st.multiselect("Select your skin concerns", ["Acne", "Dryness", "Dark Spots", "Aging", "Dullness"])

if st.button("Generate Routine"):
    if not skin_type or not concerns:
        st.error("Please select both skin type and concerns.")
    else:
        st.success("✅ Routine generated based on your profile!")

        # Simulate intelligent matching (placeholder for LLM/CrewAI logic)
        routine = {
            "cleanser": product_catalog["cleanser"][0],
            "toner": product_catalog["toner"][0],
            "moisturizer": product_catalog["moisturizer"][0],
            "serum": product_catalog["serum"][0],
            "sunscreen": product_catalog["sunscreen"][0]
        }

        st.subheader("🌿 Your Personalized Skincare Routine")
        for step, product in routine.items():
            st.write(f"**{step.capitalize()}**: {product}")
```
